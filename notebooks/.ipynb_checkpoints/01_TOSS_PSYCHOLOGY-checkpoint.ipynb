```python
# 1. Set the working directory explicitly to the notebooks folder
import os
os.chdir(r'C:\Users\91798\Documents\ANKIT SONI\Pressure-Quotient-IPL\notebooks')

# 2. Import pandas for data manipulation
import pandas as pd

# 3. Import matplotlib for creating basic charts
import matplotlib.pyplot as plt

# 4. Import seaborn for beautiful chart styling
import seaborn as sns

# 5. Tell matplotlib to display charts directly inside the Jupyter Notebook
%matplotlib inline

# 6. Load the datasets using relative paths
df_matches = pd.read_csv('../data/raw/ipl_matches_data.csv')
df_teams = pd.read_csv('../data/raw/team_aliases.csv')

# 7. Display the first 5 rows of match data to verify it loaded correctly
df_matches.head()
```

```python
# Create a new DataFrame keeping only matches where the 'result' is strictly a 'win'
df_clean = df_matches[df_matches['result'] == 'win'].copy()

# Merge the clean match data with the teams data to get the toss winner's readable name
df_clean = df_clean.merge(df_teams, left_on='toss_winner', right_on='team_id', how='left')

# Rename the newly attached 'alias_name' column to 'toss_winner_name' for clarity
df_clean = df_clean.rename(columns={'alias_name': 'toss_winner_name'})

# Drop the 'team_id' column because we already have 'toss_winner' and don't need duplicates
df_clean = df_clean.drop(columns=['team_id'])

# Merge again to get the match winner's readable name
df_clean = df_clean.merge(df_teams, left_on='match_winner', right_on='team_id', how='left')

# Rename the newly attached 'alias_name' column to 'match_winner_name'
df_clean = df_clean.rename(columns={'alias_name': 'match_winner_name'})

# Drop the duplicate 'team_id' column again
df_clean = df_clean.drop(columns=['team_id'])

# Display a few specific columns to verify our clean names are attached
df_clean[['match_id', 'season_id', 'toss_decision', 'toss_winner_name', 'match_winner_name']].head()
```

```python
# Group data by season_id and toss_decision, then count the number of occurrences
toss_counts = df_clean.groupby(['season_id', 'toss_decision']).size().unstack(fill_value=0)

# Calculate percentage of Field vs Bat decisions per season (sums across rows to 100%)
toss_pct = toss_counts.div(toss_counts.sum(axis=1), axis=0) * 100

# Round percentages to 1 decimal place for clean reporting
toss_pct = toss_pct.round(1)

# Reset index so 'season_id' becomes a standard column
toss_summary = toss_pct.reset_index()

# Display the resulting table
toss_summary
```

```python
# Set figure dimensions and resolution
plt.figure(figsize=(12, 6), dpi=100)

# Create stacked bar chart using matplotlib
plt.bar(toss_summary['season_id'], toss_summary['field'], label='Chose Field First', color='#1f77b4', width=0.7)
plt.bar(toss_summary['season_id'], toss_summary['bat'], bottom=toss_summary['field'], label='Chose Bat First', color='#d62728', width=0.7)

# Add reference line at 50% threshold
plt.axhline(50, color='gray', linestyle='--', alpha=0.7, label='50% Neutral Line')

# Customize titles and axis labels
plt.title('The Great Flip: How IPL Captains Shifted to Field-First (2008–2026)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('IPL Season', fontsize=11, labelpad=10)
plt.ylabel('Percentage of Toss Choices (%)', fontsize=11)
plt.xticks(toss_summary['season_id'], rotation=45)
plt.ylim(0, 105)
plt.legend(loc='upper left', frameon=True)

# Ensure layout is compact and clean
plt.tight_layout()

# Save image directly to the outputs/charts folder
plt.savefig('../outputs/charts/toss_01_great_flip.png', dpi=300)

# Render plot in notebook
plt.show()
```

```python
# Create a column indicating if the toss winner won the match (1 if True, 0 if False)
df_clean['toss_winner_won'] = (df_clean['toss_winner'] == df_clean['match_winner']).astype(int)

# Group by season and toss_decision to calculate win rate when choosing field vs choosing bat
wr_by_season = df_clean.groupby(['season_id', 'toss_decision'])['toss_winner_won'].mean().unstack() * 100

# Round to 1 decimal place
wr_by_season = wr_by_season.round(1).reset_index()

# Set up the figure for Chart 2
plt.figure(figsize=(12, 6), dpi=100)

# Plot line for Field First win rate
plt.plot(wr_by_season['season_id'], wr_by_season['field'], marker='o', color='#1f77b4', linewidth=2.5, label='Field First Win Rate %')

# Plot line for Bat First win rate
plt.plot(wr_by_season['season_id'], wr_by_season['bat'], marker='s', color='#d62728', linewidth=2.5, label='Bat First Win Rate %')

# Add 50% neutral reference line
plt.axhline(50, color='gray', linestyle='--', alpha=0.7, label='50% Baseline')

# Identify and highlight seasons where Batting First win rate was strictly greater than Fielding First
bat_better_seasons = wr_by_season[wr_by_season['bat'] > wr_by_season['field']]

# Add visual markers (gold star) for the seasons where the herd was wrong
plt.scatter(bat_better_seasons['season_id'], bat_better_seasons['bat'], color='gold', s=150, zorder=5, edgecolor='black', label='Bat > Field (Herd Wrong)')

# Formatting titles and labels
plt.title('The Herd Was Wrong 6 Times: Field vs Bat Win Rate by Season (2008–2026)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('IPL Season', fontsize=11, labelpad=10)
plt.ylabel('Win Rate (%)', fontsize=11)
plt.xticks(wr_by_season['season_id'], rotation=45)
plt.ylim(20, 80)
plt.legend(loc='lower left', frameon=True)

# Save chart directly to outputs folder
plt.tight_layout()
plt.savefig('../outputs/charts/toss_02_win_rate_paradox.png', dpi=300)
plt.show()
```

```python
# Group by season to calculate the percentage of times the toss winner won the match
toss_adv_season = df_clean.groupby('season_id')['toss_winner_won'].mean() * 100

# Reset index to turn it back into a standard DataFrame
toss_adv_season = toss_adv_season.reset_index()

# Calculate a 3-season rolling average to smooth out the trend (min_periods=1 prevents NaN in early years)
toss_adv_season['rolling_3yr'] = toss_adv_season['toss_winner_won'].rolling(window=3, min_periods=1).mean()

# Set up the figure for Chart 3
plt.figure(figsize=(12, 6), dpi=100)

# Create a list of colors: Green if win rate > 50%, Red if < 50%
colors = ['#2ca02c' if val > 50 else '#d62728' for val in toss_adv_season['toss_winner_won']]

# Plot the bars representing each season's toss win percentage
plt.bar(toss_adv_season['season_id'], toss_adv_season['toss_winner_won'], color=colors, alpha=0.7, label='Season Toss Win %')

# Plot the 3-season rolling average line on top of the bars
plt.plot(toss_adv_season['season_id'], toss_adv_season['rolling_3yr'], color='purple', linewidth=3, marker='o', label='3-Season Rolling Avg')

# Add a prominent 50% reference line (the pure luck baseline)
plt.axhline(50, color='black', linestyle='dashed', linewidth=2, label='Pure Luck Baseline (50%)')

# Customize titles and labels
plt.title('The Toss Illusion: Winning the Toss Barely Beats a Coin Flip', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('IPL Season', fontsize=11, labelpad=10)
plt.ylabel('% of Toss Winners Who Won the Match', fontsize=11)
plt.xticks(toss_adv_season['season_id'], rotation=45)
plt.ylim(35, 65) # Zoom in on the 35-65 range to show the minor fluctuations
plt.legend(loc='upper right', frameon=True)

# Ensure layout is compact and clean
plt.tight_layout()

# Save image directly to the outputs/charts folder
plt.savefig('../outputs/charts/toss_03_toss_illusion.png', dpi=300)

# Render plot in notebook
plt.show()
```

```python
# Filter matches won by wickets (chasing) and matches won by runs (defending)
win_wickets = df_clean[df_clean['win_by_wickets'] > 0]['win_by_wickets']
win_runs = df_clean[df_clean['win_by_runs'] > 0]['win_by_runs']

# Create a figure with 2 side-by-side subplots
fig, axes = plt.subplots(1, 2, figsize=(12, 5), dpi=100)

# Subplot 1: Chasing Win Margin (Wickets)
sns.boxplot(y=win_wickets, ax=axes[0], color='#1f77b4', width=0.35)
axes[0].set_title('Chasing Victory Margin (Wickets)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Wickets Remaining', fontsize=11)
axes[0].axhline(win_wickets.median(), color='darkblue', linestyle='--', label=f'Median: {win_wickets.median():.1f} Wickets')
axes[0].legend(loc='upper right')

# Subplot 2: Defending Win Margin (Runs)
sns.boxplot(y=win_runs, ax=axes[1], color='#d62728', width=0.35)
axes[1].set_title('Defending Victory Margin (Runs)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Runs Won By', fontsize=11)
axes[1].axhline(win_runs.median(), color='darkred', linestyle='--', label=f'Median: {win_runs.median():.1f} Runs')
axes[1].legend(loc='upper right')

# Add overall title
plt.suptitle('Win Margin Psychology: Why Chasing Feels Easier Than Defending', fontsize=14, fontweight='bold', y=1.03)

# Adjust layout and save plot directly
plt.tight_layout()
plt.savefig('../outputs/charts/toss_04_win_margin_psychology.png', dpi=300)
plt.show()
```

```python
# 1. Identify the top 15 venues by total matches played
top_15_venues = df_clean['venue'].value_counts().head(15).index
df_top_venues = df_clean[df_clean['venue'].isin(top_15_venues)].copy()

# 2. Build a summary table for these venues
venue_stats = pd.DataFrame()
venue_stats['total_matches'] = df_top_venues.groupby('venue').size()
venue_stats['field_choices'] = df_top_venues[df_top_venues['toss_decision'] == 'field'].groupby('venue').size()

# Calculate Field Choice %
venue_stats['field_choice_pct'] = (venue_stats['field_choices'] / venue_stats['total_matches']) * 100

# 3. Calculate actual Win % when fielding first at each venue
field_first_matches = df_top_venues[df_top_venues['toss_decision'] == 'field']
venue_stats['field_wins'] = field_first_matches.groupby('venue')['toss_winner_won'].sum()
venue_stats['field_win_pct'] = (venue_stats['field_wins'] / venue_stats['field_choices']) * 100

# Clean up, reset index, and sort by choice percentage for a neat chart
venue_stats = venue_stats.reset_index().sort_values('field_choice_pct', ascending=True)

# 4. Define dynamic colors based on actual win percentage
def assign_color(win_pct):
    if win_pct > 55: return '#2ca02c' # Green: Good choice (Win > 55%)
    elif win_pct >= 45: return '#ff7f0e' # Orange: Neutral (Win 45-55%)
    else: return '#d62728' # Red: Bad choice (Win < 45%)

venue_stats['color'] = venue_stats['field_win_pct'].apply(assign_color)

# 5. Plotting the Horizontal Bar Chart
plt.figure(figsize=(12, 8), dpi=100)

plt.barh(venue_stats['venue'], venue_stats['field_choice_pct'], color=venue_stats['color'], alpha=0.8)

# Add reference lines
plt.axvline(50, color='gray', linestyle='--', label='50% Neutral Choice')
overall_avg = (df_clean['toss_decision'] == 'field').mean() * 100
plt.axvline(overall_avg, color='black', linestyle=':', linewidth=2, label=f'IPL Average Field Choice ({overall_avg:.0f}%)')

# Customize titles and labels
plt.title('Venue Psychology: Is the Obsession with Fielding First Justified?', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('% of Matches Captain Chose to Field First', fontsize=11)
plt.ylabel('Top 15 Venues', fontsize=11)

# Create custom legend for the colors
import matplotlib.patches as mpatches
green_patch = mpatches.Patch(color='#2ca02c', label='Fielding Wins > 55% (Justified)')
orange_patch = mpatches.Patch(color='#ff7f0e', label='Fielding Wins 45-55% (Coin Flip)')
red_patch = mpatches.Patch(color='#d62728', label='Fielding Wins < 45% (Unjustified)')
plt.legend(handles=[green_patch, orange_patch, red_patch], loc='lower right', frameon=True)

# Save directly to outputs folder
plt.tight_layout()
plt.savefig('../outputs/charts/toss_05_venue_psychology.png', dpi=300)
plt.show()
```

```python
# Create a 2x2 grid for our infographic
fig, axes = plt.subplots(2, 2, figsize=(14, 10), dpi=100)
fig.suptitle('4 Cognitive Biases Behind the Field-First Orthodoxy in IPL', fontsize=18, fontweight='bold', y=0.96)

# Panel 1: Availability Heuristic (Text + Mini Bar Chart)
axes[0, 0].set_title('1. Availability Heuristic', fontsize=14, fontweight='bold', pad=10)
axes[0, 0].text(0.5, 0.85, 'Dramatic chases stick in memory more\nthan dull bowling victories.', ha='center', fontsize=12)
# Calculate close finishes
close_runs = len(df_clean[(df_clean['win_by_runs'] > 0) & (df_clean['win_by_runs'] <= 10)])
close_wkts = len(df_clean[(df_clean['win_by_wickets'] > 0) & (df_clean['win_by_wickets'] <= 2)])
# Plot mini bar chart
axes[0, 0].bar(['Close Defenses\n(<10 runs)', 'Close Chases\n(<2 wkts)'], [close_runs, close_wkts], color=['#d62728', '#1f77b4'], width=0.4)
axes[0, 0].set_ylim(0, max(close_runs, close_wkts) * 1.5)

# Panel 2: Regret Aversion (Text Box)
axes[0, 1].set_title('2. Regret Aversion', fontsize=14, fontweight='bold', pad=10)
axes[0, 1].text(0.5, 0.5, 'Captains make decisions to minimize regret.\n\nDefending a small total concentrates blame\non the captain. Fielding first distributes\nblame to the bowlers.\n\nAverage chasing win feels dominant (6.1 wkts)\nAverage defending win feels narrow (29.3 runs)', 
                ha='center', va='center', fontsize=13, bbox=dict(facecolor='#f8f9fa', alpha=0.8, boxstyle='round,pad=1.5'))
axes[0, 1].axis('off') # Hide axes for a clean text look

# Panel 3: Herd Mentality (Mini Line Chart)
axes[1, 0].set_title('3. Herd Mentality', fontsize=14, fontweight='bold', pad=10)
axes[1, 0].text(0.5, 0.9, 'Following the majority under uncertainty.', ha='center', fontsize=12)
axes[1, 0].plot(toss_summary['season_id'], toss_summary['field'], color='#1f77b4', linewidth=3)
axes[1, 0].axhline(50, color='gray', linestyle='--', label='50% Line')
axes[1, 0].set_xticks([2008, 2017, 2026])
axes[1, 0].set_ylabel('Fielding Preference %')

# Panel 4: Loss Aversion (Quote Box)
axes[1, 1].set_title('4. Loss Aversion', fontsize=14, fontweight='bold', pad=10)
axes[1, 1].text(0.5, 0.5, '"Losing hurts twice as much as gaining feels good."\n- Kahneman & Tversky\n\nWhen setting a target, every run conceded\nfeels like a loss. Chasing converts an\nuncertain defense into a clear, psychologically\nsafe target.', 
                ha='center', va='center', fontsize=13, style='italic', bbox=dict(facecolor='#e6f2ff', alpha=0.8, boxstyle='round,pad=1.5'))
axes[1, 1].axis('off')

# Ensure layout is neat and save
plt.tight_layout(rect=[0, 0.03, 1, 0.92])
plt.savefig('../outputs/charts/toss_06_behavioral_framework.png', dpi=300)
plt.show()
```
